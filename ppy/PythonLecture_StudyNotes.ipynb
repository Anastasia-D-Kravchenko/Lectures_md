## 1.2 Objects and References ⭐ (Very Important!)

In Python, **variable names are references (pointers) to objects**, not the objects themselves.

- `id(x)` returns the unique identity (memory address) of the object `x` points to
- `type(x)` returns the type of the object
- `is` compares *identity* (same object?); `==` compares *value*
- **Immutable types**: `int`, `float`, `str`, `tuple` — changing value creates a NEW object
- **Mutable types**: `list`, `dict`, `set` — can be modified in place, same `id`

````python
# --- Immutability of numbers ---
x = 3
print('id(x) before:', id(x))
x = x + 1        # creates a NEW int object
print('id(x) after: ', id(x))  # DIFFERENT id!

# --- Two names, one object ---
x = 7; y = 7     # small integers are cached
print('x is y:', x is y)   # True — same cached object
print('id(x):', id(x), 'id(y):', id(y))

# --- Mutable list: same id after modification ---
lst = [1, 2, 3]
id_before = id(lst)
lst.append(4)    # modifies in place
print('Same list object?', id(lst) == id_before)  # True

# --- Shallow vs deep copy ---
from copy import deepcopy
a = [1, 2, [3, 4]]
b = a            # same object! b IS a
c = list(a)      # shallow copy — outer list is new, but inner list is shared
d = deepcopy(a)  # deep copy — completely independent

a[2][0] = 99     # modify inner list
print('b (alias):', b)       # [1, 2, [99, 4]]  — affected!
print('c (shallow):', c)     # [1, 2, [99, 4]]  — inner list shared, affected!
print('d (deep):', d)        # [1, 2, [3, 4]]   — independent, NOT affected
````

## 2.1 `pass` statement
`pass` does nothing — used when syntax requires a statement but you don't need any action.
Useful for empty class/function bodies, or placeholder code.

````python
# Custom exception classes with pass
class ZeroArg(Exception): pass
class NegArg(Exception):  pass

from math import log10
def f(num, denom):
    if denom == 0: raise ZeroArg
    elif num == 0: raise NegArg
    return log10(abs(num)) / denom

for n, d in [(100, 2), (0, 2), (10, 0)]:
    try:
        print(f'f({n},{d}) =', f(n, d))
    except Exception as e:
        print(f'f({n},{d}) -> {e.__class__.__name__}')
````

## 2.2 `if` statement and conditional expression
- Conditions do **not** need parentheses
- Blocks are marked by **indentation only** (4 spaces)
- Ternary/conditional expression: `expr_if_true if condition else expr_if_false`

````python
# if / elif / else
number = 42
if number < 0:
    print('negative')
elif number == 0:
    print('zero')
elif number == 1:
    print('one')
else:
    print('many')  # prints: many

# Conditional expression (ternary)
x = 7
label = 'even' if x % 2 == 0 else 'odd'
print(f'{x} is {label}')    # 7 is odd

# Nested conditional — for traffic light example
for t in (20, 50, 90):
    colour = 'green' if t < 40 else 'yellow' if t < 80 else 'red'
    print(f't={t} -> {colour}')

# Collatz (hailstone) sequence using conditional expression
n = p = 7
seq = [n]
while p > 1:
    p, n = n, n // 2 if n % 2 == 0 else 3 * n + 1
    seq.append(n)
print('Collatz from 7:', seq)
````

## 2.3 `while` loop — with `break`, `continue`, `else`
Key rule: **`else` on a loop runs only if the loop finished WITHOUT `break`**

````python
# while with break/continue/else
def is_prime_while(n):
    if n < 2: return False
    if n in (2, 3): return True
    if n % 2 == 0: return False
    i = 3
    while i * i <= n:
        if n % i == 0:
            break          # found a divisor
        i += 2
    else:
        return True        # loop finished without break → prime
    return False

for num in [2, 7, 9, 13, 15, 17, 25]:
    print(f'{num:2d}: {"prime" if is_prime_while(num) else "not prime"}')

# while else — reading until 0
inputs = [5, 3, 8, 0]   # simulating input
idx = 0
total = 0
while True:
    n = inputs[idx]; idx += 1
    if n == 0:
        break
    total += n
else:
    print('Loop ended normally')  # won't print, we used break
print('Sum:', total)
````

## 2.4 `for` loop — iterables, iterators, `enumerate`, `range`

The `for` loop is really a `while` loop that calls `iter()` then `next()` repeatedly.

- **Iterable**: has `__iter__()` method → returns an iterator
- **Iterator**: has `__next__()` method → returns next value or raises `StopIteration`
- `enumerate(iterable, start=0)` → yields `(index, value)` tuples
- `range(start, stop, step)` → immutable sequence of integers

⚠️ **Do NOT modify a list while iterating over it** — iterate over a copy instead: `for n in lst[:]:`

````python
# What for-loop does under the hood
lst = [10, 20, 30]
it = iter(lst)          # calls lst.__iter__()
while True:
    try:
        val = next(it)  # calls it.__next__()
        print(val, end=' ')
    except StopIteration:
        break
print()

# enumerate — with start index
seasons = ['Spring', 'Summer', 'Autumn', 'Winter']
for i, s in enumerate(seasons, start=1):
    print(f'Season {i}: {s}')

# Unpacking tuples in for loop
persons = [('Ann', 23), ('Bea', 34), ('Sue', 29)]
for name, age in persons:
    print(f'{name} is {age} years old')

# Dictionary iteration
scores = {'Alice': 88, 'Bob': 95, 'Carol': 72}
for key, value in scores.items():
    print(f'{key}: {value}')

# range examples
print(list(range(5)))           # [0,1,2,3,4]
print(list(range(1, 11)))       # [1..10]
print(list(range(10, 0, -1)))   # [10..1]
print(list(range(24, -2, -5)))  # [24,19,14,9,4,-1]
````

## 2.5 `match` statement (Python 3.10+)

Like `switch` in Java/C++, but **much more powerful** — can match:
- literal values (`case 1:`, `case 'hello':`) with `|` for OR
- sequences of different lengths (`case [a, b]:`, `case [c, *d]:`) 
- types (`case int() as a:`, `case str() as s:`)
- type + guards (`case int() as a if a < 0:`)
- `case _:` is the catch-all (like `default:`)
- **No fall-through** — only first matching case executes

````python
# Basic match — like switch
k = 3
match k:
    case 0:       print('zero')
    case 1|2|3|4: print('less than 5')   # OR with |
    case 5:       print('five')
    case _:       print('other')

# Match sequences of different lengths
items = [(1, 2, 3), ['a', 'b'], (4, 5, 6, 7, 8), 101]
for item in items:
    match item:
        case [a1, a2]:      print(f'Two elements: {a1}, {a2}')
        case [b1, b2, b3]:  print(f'Three elements: {b1}, {b2}, {b3}')
        case [c, *d]:       print(f'Many elements: {c}, rest={d}')
        case e:             print(f'Single value: {e}')

# Match by type with guard
subjects = [{}, 8, 'Warsaw', [1,2], -2.0, 0, -1, 'Dog']
for sub in subjects:
    match sub:
        case int()|float() as a if a < 0: print(f'Negative number: {a}')
        case int()|float() as a if a == 0: print('Zero')
        case int()|float() as a:          print(f'Positive number: {a}')
        case str() as w if len(w) > 5:    print(f'Long word: {w}')
        case str() as w:                  print(f'Short word: {w}')
        case w:                           print(f'Other: {w} type={type(w).__name__}')
````

## 3.2 Boolean type

**Falsy** values (evaluate to `False` in boolean context):
- `0`, `0.0`, `0j` (zero numbers)
- `''`, `[]`, `{}`, `set()`, `()` (empty sequences/collections)
- `None`, `False`

Everything else is **truthy**.

**Important**: `bool` inherits from `int` → `True == 1`, `False == 0`

Logical operators: `and`, `or`, `not` (NOT `&&`, `||`, `!`)

⭐ `and`/`or` return **one of their operands**, not necessarily `True`/`False`:
- `expr1 and expr2` → returns `expr1` if falsy, else `expr2`
- `expr1 or expr2` → returns `expr1` if truthy, else `expr2`

````python
# Falsy / Truthy
falsy = [0, 0.0, '', [], {}, set(), None, False]
for v in falsy:
    print(f'  bool({repr(v):12}) = {bool(v)}')

print()
truthy = [1, -1, 0.001, 'hi', [0], {'a':1}, True]
for v in truthy:
    print(f'  bool({repr(v):12}) = {bool(v)}')

print()
# and / or return an OPERAND, not True/False
print('0 and [1,2]       =', 0 and [1, 2])         # 0   (falsy first)
print('-1 and [1,2]      =', -1 and [1, 2])        # [1,2] (truthy first)
print('[] or {1,2}       =', [] or {1, 2})         # {1,2} (first is falsy)
print('[3] or {1,2}      =', [3] or {1, 2})        # [3]   (first is truthy)

# XOR using bool: bool(a) ^ bool(b)
a, b = True, False
print('XOR:', bool(a) ^ bool(b))  # True

# bool inherits from int — sometimes surprising
print(True + True)    # 2
print(True == 1)      # True
print(False == 0)     # True
````

## 3.3 Integer (`int`)
- **Unlimited precision** — Python integers can be arbitrarily large
- Small integers `[-5, 256]` are **cached** (same object for all references)
- Literals: `0b1010` (binary), `0o17` (octal), `0x2F` (hex), `1_000_000` (underscores ok)
- `bin(n)`, `oct(n)`, `hex(n)` → string representations

````python
# Unlimited precision
big = 2 ** 200
print(big)
print(f'Digits: {len(str(big))}')

# Small integer caching
a = 256; b = 256
print('256 is cached:', a is b)   # True
a = 257; b = 257
print('257 is cached:', a is b)   # False — not cached

# Literals in different bases
print(47 == 0b101111 == 0o56 == 0x2F)  # True
print(1_234_567)  # underscore separators for readability

# Conversion functions
print(bin(255))   # '0b11111111'
print(oct(255))   # '0o377'
print(hex(255))   # '0xff'

# From string to int with base
print(int('1111', 2))   # 15  (binary)
print(int('FF', 16))    # 255 (hex)
print(int('0xFF', 16))  # 255 (with prefix)
print(int('57', 8))     # 47  (octal)
print(int('2Z', 36))    # 107 (base 36!)

# Memory size of integers
import sys
for n in [1, 2**30-1, 2**30, 2**60]:
    print(f'{n}: size = {sys.getsizeof(n)} bytes')
````

## 3.4 Float (`float`) — IEEE 754 double
- Corresponds to `double` in C/C++/Java — 64 bits, ~15-17 significant digits
- **⚠️ Not all decimal numbers are represented exactly** — only powers-of-2 fractions
- Special values: `float('inf')`, `float('-inf')`, `float('nan')`
- `NaN != NaN` is always `True` — NaN compared with anything gives `False`

````python
# Floating point precision issue
import decimal
print(decimal.Decimal(0.1))  # 0.10000000000000000555...

total = sum(decimal.Decimal('0.1') for _ in range(1_000_000))
print(f'Sum of 1M * 0.1 = {total}')           # not exactly 100000
print(f'Error: {abs(total - 100_000):.2e}')   # small but nonzero

# 0.125 = 1/8 is exact
total2 = sum(0.125 for _ in range(1_000_000))
print(f'Sum of 1M * 0.125 = {total2}')        # exactly 125000.0

# Special float values
inf = float('inf')
nan = float('nan')
print(inf + 100, inf - inf, nan == nan)  # inf nan False

# int() truncates towards zero (NOT floor)
print(int(3.9))   # 3   (NOT 4)
print(int(-3.9))  # -3  (NOT -4!)
import math
print(math.floor(-3.9))  # -4  (floor rounds towards -inf)
````

## 3.5 Complex numbers and Strings

**Complex**: written with `j` (not `i`): `3+4j`, `1j`, `0.5*(1+1j)**2`

**Strings** (`str`) are **immutable** sequences of Unicode characters:

## 4.1 Assignment — NOT an operator in Python

- `=` is a **statement**, not an expression — it has no value (`a = (b = 2)` is a SyntaxError)
- **Chained assignment**: `a = b = c = 5` — evaluates rightmost expression once, then assigns left-to-right
- **Walrus operator** `:=` (Python 3.8+) — assigns AND returns value
  - Must usually be in parentheses
  - Only works with simple variable names (not `lst[0] := x`)
  - The name comes from its appearance resembling walrus eyes

````python
# Chained assignment
a = b = c = 42
print(a, b, c)  # 42 42 42
print(a is b is c)  # True — all reference the same object

# Walrus operator := (assignment expression)
# Without walrus — input repeated:
inputs = ['hello', 'Python', 'world', 'quit']
idx = 0
words = []
while (word := inputs[idx]) != 'quit':
    words.append(word.upper())
    idx += 1
print('Words:', words)

# Walrus in if — avoiding recalculation
s = 'rotator'
if (r := s[::-1]) == s:
    print(f'{s!r} is a palindrome; reversed = {r!r}')

# Walrus in comprehension — avoid recalculating fib
def fib(n):
    if n < 2: return n
    return fib(n-1) + fib(n-2)

odd_fibs = [(n, r) for n in range(20, 25) if (r := fib(n)) % 2 != 0]
print('Odd Fibonacci:', odd_fibs)
````

## 4.3 Division and Modulo — Key Python Difference!

| Operator | Python | C/C++/Java |
|----------|--------|------------|
| `9 / 4` | `2.25` (float!) | `2` (integer) |
| `-5 // 4` | `-2` (floor: towards -∞) | `-1` (truncate: towards 0) |
| `17 % -10` | `-3` | `7` |

Python rule: `a % b = a - (a // b) * b`

````python
# / always returns float
print(9 / 3)     # 3.0  NOT 3!
print(9 / 4)     # 2.25
print(1 / 3)     # 0.333...
print(type(9/3)) # <class 'float'>

# // is floor division (towards -infinity)
print(5 // 4)    #  1  (same as C)
print(-5 // 4)   # -2  (DIFFERENT from C: C gives -1)
print(17 // 10)  #  1
print(-17 // 10) # -2  (floor of -1.7)

# Modulo — always satisfies: a == (a//b)*b + a%b
for a, b in [(17,10), (17,-10), (-17,10), (-17,-10)]:
    r = a % b
    check = (a // b) * b + r
    print(f'{a:3} % {b:3} = {r:3}   check: {check} == {a}')

# Exponentiation — ** not ^
print(2**10)     # 1024
print(2**-3)     # 0.125
print(-2**4)     # -16 (means -(2**4), NOT (-2)**4!)
print((-2)**4)   # 16
print(pow(2, 3, 5))  # 8 mod 5 = 3  (three-arg pow)
print(8%5)
````

## 4.4 Comparisons

- **Chained comparisons**: `a < b < c` means `a < b and b < c` (b evaluated once!)
- **`==`** compares values; **`is`** compares identity (same object)
- **`in`** / **`not in`** — membership test (searches keys in dicts)
- Small integers `[-5, 256]` are cached — `is` may return `True` unexpectedly

````python
# Chained comparisons — unique to Python
x = 5
print(0 < x < 10)      # True
print(1 < 2 < 3 < 4)   # True
print(5 < 7 > 2)       # True  (5<7 and 7>2)
print(5 < 7 > 7)       # False (7>7 is False)

# is vs == 
a = 'Hello, '
b = 'World'
s1 = a + b
s2 = 'Hello, World'
print('s1 == s2:', s1 == s2)   # True  (same value)
print('s1 is s2:', s1 is s2)   # False (different objects)

# Small int caching
a = 17; b = 17
print('17 is 17:', a is b)      # True (cached!)
a = 1200; b = 1200
print('1200 is 1200:', a is b)  # False (not cached)

# Membership tests
lst = [(1,2), (3,4,5)]
dct = {'one': 1, 'two': 2}
st  = 'abcde'
print((1,2) in lst)    # True
print('one' in dct)    # True (checks KEYS)
print(1 in dct)        # False (1 is not a key)
print('cd' in st)      # True (substring check)
````

## 4.5 Indexing and Slicing

- Index 0 is first, -1 is last, -n is n-th from the end
- `seq[start:stop:step]` — stop is **excluded**
- For positive step: `start` defaults to 0, `stop` defaults to length
- For **negative step**: `start` defaults to -1, `stop` defaults to `-len-1`
- `seq[::-1]` → reversed copy
- `seq[:]` → shallow copy
- Slices return **shallow copies** — nested mutable objects are shared!

````python
lst = ['a','b','c','d','e','f']
#       0   1   2   3   4   5
#      -6  -5  -4  -3  -2  -1

print(lst[0], lst[-1])   # a f
print(lst[:3])            # ['a','b','c']
print(lst[3:])            # ['d','e','f']
print(lst[1::2])          # ['b','d','f']  every 2nd from index 1
print(lst[::-1])          # ['f','e','d','c','b','a']  reversed
print(lst[5:1:-1])        # ['f','e','d','c']
print(lst[1:-2])          # ['b','c','d','e']  without first and last

# Slicing strings
s = 'PYTHON'
print(s[::2])             # 'PTO' (every 2nd)
print(s[::-1])            # 'NOHTYP' (reversed)

# Modify list via slice
a = [1, 2, 3, 4, 5]
a[1:4] = [20, 30]    # replace 3 elements with 2
print(a)              # [1, 20, 30, 5]
a[1:3] = []          # delete elements
print(a)              # [1, 5]

# index() and find()
b = [5, 6, 7, 5, 6, 7]
print(b.index(5))        # 0  (first occurrence)
print(b.index(5, 2))     # 3  (search from index 2)
s2 = 'abcdefg'
print(s2.find('d'))      # 3
print(s2.find('z'))      # -1  (not found, no exception)
````

# Chapter 5 — Functions

## 5.1 Functions as Objects + Scopes (LEGB Rule)

Functions are **objects** — they can be assigned to variables, passed as arguments, returned from other functions.

**LEGB rule** — Python searches names in this order:
1. **L**ocal — inside current function
2. **E**nclosing — surrounding function (for nested functions)
3. **G**lobal — module level
4. **B**uilt-in — Python built-ins

- `global x` — declare that `x` in the function refers to the module-level variable
- `nonlocal x` — declare that `x` refers to the enclosing function's variable
- **Closures**: inner functions 'remember' variables from their enclosing scope

````python
# Functions are objects
def adder(a, b):
    """Return the sum of a and b."""
    return a + b

print(type(adder))         # <class 'function'>
print(adder.__name__)      # adder
print(adder.__doc__)       # Return the sum...

add = adder                # copy the reference
print(add is adder)        # True — same object
print(add(3, 4))           # 7

# Duck typing — same function works for any + compatible type
print(adder(1, 2))         # 3 (ints)
print(adder('Hello', ' World'))  # 'Hello World' (strings!)
print(adder([1,2], [3,4])) # [1,2,3,4] (lists!)

print()
# LEGB + global / nonlocal
n = 100  # global

def outer():
    n = 10  # enclosing (local to outer)
    def inner():
        nonlocal n        # refers to outer's n
        n *= 2
        return n
    return inner

f = outer()
print(f())    # 20   (outer's n was 10 → 20)
print(f())    # 40   (closure remembers updated value)
print(n)      # 100  (global unchanged)

# Counter using closure
def make_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter

c = make_counter()
print(c(), c(), c())  # 1 2 3
````

## 5.2 Passing Arguments

- **Positional**: in order, no name
- **Keyword**: `f(c=3, a=1, b=2)` — order doesn't matter
- **Default values**: `def f(a, b='default')` — evaluated ONCE at definition time (⚠️ mutable defaults!)
- **`*args`**: variadic positional → collected into a **tuple**
- **`**kwargs`**: variadic keyword → collected into a **dict**
- **`/`**: positional-only parameters before `/`
- **`*`**: keyword-only parameters after `*`
- Unpacking: `f(*lst)` unpacks list into positional args; `f(**dct)` unpacks dict into keyword args

````python
# Basic argument types
def f(a, b, c):
    return f'{a=} {b=} {c=}'

print(f(1, 'Hello', 2.5))           # positional
print(f(c='World', a=17, b=(3,'3'))) # keyword (any order)
print(f(21, c=8.28, b='Ottawa'))     # mixed

# Unpacking * and **
lst = ['a', 1, 'Hello']
print(f(*lst))                        # unpack list
dct = dict(b='Hello', c=1.25)
print(f('First', **dct))              # unpack dict

# Default values — ⚠️ created ONCE at definition time
i = 7
def fundef(a, b=i):
    return a + b
i += 10
print(fundef(1))  # 8 — default is still 7, not 17!

# ⚠️ Mutable default — the list persists between calls!
def bad_append(val, lst=[]):
    lst.append(val)
    return lst
bad_append(1); bad_append(2); bad_append(3)
print(bad_append(4))  # [1,2,3,4] — not empty!

# Fix: use None as default
def good_append(val, lst=None):
    if lst is None: lst = []
    lst.append(val)
    return lst
a = good_append()
print(a(3))
good_append(1); good_append(2); good_append(3)
print(good_append(4))

print()
# *args and **kwargs
def describe(*args, **kwargs):
    print('Traits:', ', '.join(args))
    for k, v in kwargs.items():
        print(f'  {k}: {v}')

describe('tall', 'smart', name='Jan', age=30)

# Positional-only (/) and keyword-only (*)
def strict(pos_only, /, normal, *, kw_only):
    print(pos_only, normal, kw_only)

strict(1, 2, kw_only=3)         # works
strict(1, normal=2, kw_only=3)  # also works
# strict(pos_only=1, normal=2, kw_only=3)  # ERROR!
````

## 5.3 Lambda Expressions

`lambda args: expr` — anonymous function, body must be a **single expression**

- Type is `function` — same as `def`
- Can use conditional expressions: `lambda x: 'even' if x%2==0 else 'odd'`
- Can be recursive (must have a name)
- Used with `sorted(key=...)`, `filter(pred, ...)`, `map(func, ...)`

````python
# Lambda basics
sq = lambda x: x * x
add = lambda x, y: x + y
pal = lambda s: 'palindrome' if s == s[::-1] else 'not palindrome'

print(sq(7), add(3, 4), pal('rotor'), pal('rover'))

# Lambda as argument to sorted()
people = [('Joe', 182, 79), ('Sue', 173, 59), ('Liz', 175, 62)]
by_weight = sorted(people, key=lambda p: p[2])
by_height = sorted(people, key=lambda p: p[1])
print('By weight:', by_weight)
print('By height:', by_height)

# Sort with multiple criteria (tuple key)
words = ['jane', 'Bertha', 'henry', 'alice', 'Hugh']
result = sorted(words, key=lambda s: (len(s), s.lower()))
print('By length then alpha:', result)

# Lambda returning lambda — composition
cmp = lambda f, g: lambda x: f(g(x))
sq_then_add1 = cmp(lambda x: x+1, lambda x: x**2)
add1_then_sq = cmp(lambda x: x**2, lambda x: x+1)
print(sq_then_add1(3), add1_then_sq(3))  # 10, 16

# Recursive lambda (must have a name before use)
fac = lambda n: 1 if n <= 1 else n * fac(n-1)
print([fac(n) for n in range(1, 8)])
````

## 5.4 `filter`, `map`, `reduce`, `accumulate`, `sorted`

- **`filter(pred, iterable)`** → iterable of elements where `pred` returns truthy; `pred=None` keeps truthy elements
- **`map(func, iter1, iter2...)`** → applies `func` to each element (or tuple of elements)
- **`reduce(func, iterable [,init])`** (from `functools`) → folds iterable to single value
- **`accumulate(iterable, func)`** (from `itertools`) → running totals (all intermediate results)
- **`sorted(iterable, key=None, reverse=False)`** → new sorted list; `key` applied for comparison

````python
# filter
lst = [1, 5, 1, 0, '', {}, 2]
print('Truthy only:', list(filter(None, lst)))   # [1, 5, 1, 2]

names = ['Kyle', 'Sue', 'Mary', 'Ada']
print('Long names:', list(filter(lambda s: len(s) > 3, names)))  # Kyle Mary

# Filter a dictionary
ages = {'Alice': 15, 'Jane': 21, 'Kyle': 12, 'Mary': 28}
adults = dict(filter(lambda kv: kv[1] >= 18, ages.items()))
print('Adults:', adults)

# map
nums = [1, 4, 7, 2, 6]
print('Squared-1:', list(map(lambda x: (x-1)**2, nums)))

# map with 2 iterables — stops at shortest
ints   = [1, 3, 2]
strs   = ['a', 'bc', 'def', 'ghij']
print('Repeated:', list(map(lambda n,s: n*s, ints, strs)))

# reduce
from functools import reduce
nums2 = [3, 6, -1, 3, 7, 2]
total = reduce(lambda x,y: x+y, nums2)
mn    = reduce(lambda x,y: min(x,y), nums2)
mx    = reduce(lambda x,y: max(x,y), nums2)
print(f'Sum={total}, Min={mn}, Max={mx}')

# accumulate — running totals
from itertools import accumulate
payments = [150, 200, 300, 350]
# savings account: 5% interest then add payment
balance = list(accumulate(payments, lambda acc, e: round(1.05*acc + e, 1)))
print('Account balance after each payment:', balance)

# sorted with key
words2 = ['Jane', 'Hugh', 'bertha', 'henry', 'alice']
print('Default sort:       ', sorted(words2))
print('Case-insensitive:   ', sorted(words2, key=str.lower))
print('By length:          ', sorted(words2, key=len))
print('Reversed by length: ', sorted(words2, key=len, reverse=True))
````

## 5.5 Decorators

A **decorator** is a callable that takes a function and returns a (usually modified) function.

```python
@decorator
def func(): ...
```
is exactly equivalent to:
```python
def func(): ...
func = decorator(func)
```

- Use `@functools.wraps(func)` inside decorator to preserve `__name__`, `__doc__`, etc.
- `@functools.wraps` also adds `__wrapped__` attribute → access original undecorated function
- **Nested decorators**: `@italic @strong def f` means `f = italic(strong(f))`
- **Decorator factory**: `@rangecheck(0,255)` → `func = rangecheck(0,255)(func)`
- Decorators run **at module load time** (when the decorator line is processed)

````python
import functools

# 1. Simple decorator — counting invocations
def counting(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)
    wrapper.calls = 0
    return wrapper

@counting
def adder(a, b):
    """Return sum."""
    return a + b

print(adder(1, 2), adder(5, 7))
print('Calls:', adder.calls)        # 2
print('Name:', adder.__name__)      # adder (preserved by wraps!)
print('Doc:', adder.__doc__)        # Return sum.

print()
# 2. Memoization decorator
def memo(func):
    @functools.wraps(func)
    def wrapper(*args):
        if args not in wrapper.cache:
            wrapper.cache[args] = func(*args)
        return wrapper.cache[args]
    wrapper.cache = {}
    return wrapper

@memo
def fibo(n):
    if n < 2: return n
    return fibo(n-1) + fibo(n-2)

print([fibo(n) for n in range(15)])  # instant with memo

print()
# 3. Nested decorators
def italic(func):
    @functools.wraps(func)
    def _f(s): return '<i>' + func(s) + '</i>'
    return _f

def strong(func):
    @functools.wraps(func)
    def _f(s): return '<strong>' + func(s) + '</strong>'
    return _f

@italic
@strong
def decorated(s):
    return s

print(decorated(' hello '))   # <i><strong> hello </strong></i>

print()
# 4. Decorator factory with arguments
def rangecheck(mn, mx):
    def checker(func):
        @functools.wraps(func)
        def wrap(*args):
            for a in args:
                if a < mn or a > mx:
                    raise ValueError(f'Arg {a} out of range [{mn},{mx}]')
            return func(*args)
        return wrap
    return checker

@rangecheck(0, 255)
def rgb(r, g, b):
    return f'rgb({r},{g},{b})'

print(rgb(100, 200, 50))
try:
    print(rgb(100, 256, 50))   # 256 out of range
except ValueError as e:
    print(e)
````

## 5.6 Single Dispatch (function 'overloading')

`@functools.singledispatch` creates a generic function with versions registered per type.
The correct version is selected based on the **type of the first argument**.
Subtypes use their parent's implementation if not separately registered.

````python
import functools

@functools.singledispatch
def process(arg):
    raise NotImplementedError(f'No handler for {type(arg).__name__}')

@process.register(int)
def _(v):
    print(f'Processing int: {v}, squared = {v**2}')

@process.register(str)
def _(v):
    print(f'Processing str: {v!r}, upper = {v.upper()}')

@process.register(list)
def _(v):
    print(f'Processing list of {len(v)}: sum = {sum(x for x in v if isinstance(x, (int,float)))}')

process(42)
process('hello')
process([1, 2, 3, 4])
process(True)    # bool inherits int → uses int handler!

try:
    process(3.14)  # float not registered
except NotImplementedError as e:
    print(e)

# See registered types
print('Registered:', list(process.registry.keys()))
````

## 6.1 List Comprehensions

```python
[expr for var in iterable]
[expr for var in iterable if condition]
[expr for x in outer for y in inner if cond1 if cond2]
```
- The loop variable is **local** to the comprehension (doesn't leak)
- Nested: `[k for v in ls if isinstance(v,list) for k in v if k%2==0]`
  matches the **equivalent nested for-loops** top-to-bottom

````python
# Examples from the lecture

# Squares
print([x*x for x in range(1, 11)])

# Squares of even numbers only
print([x*x for x in range(1, 11) if x % 2 == 0])

# Different operations for odd/even
print([x*x if x%2 else x*x*x for x in range(1, 7)])

# Selecting by type
mixed = [1, 'a', 'bc', 1.2e2, True, None]
print('Numbers:', [x for x in mixed if isinstance(x, int|float)])
print('Strings:', [x for x in mixed if isinstance(x, str)])

# Nested — flatten a 2D list
matrix = [[1,2,3], [4,5,6], [7,8,9]]
flat = [k for row in matrix for k in row]
print('Flattened:', flat)

# Nested — only even numbers from lists inside a list
ls = [[1,2], 'abc', [3,4,5,6], 'xy']
evens = [k for v in ls if isinstance(v, list) for k in v if k%2==0]
print('Evens from inner lists:', evens)

# Prime numbers ≤ 31
from math import isqrt
primes = [2] + [k for k in range(3, 32, 2)
                if all(k%m != 0 for m in range(2, isqrt(k)+1))]
print('Primes ≤ 31:', primes)

# Convert 2D string matrix to ints
str_matrix = [['4','8'], ['4','2','28'], ['1','12']]
int_matrix = [[int(e) for e in row] for row in str_matrix]
print('Int matrix:', int_matrix)

# Remove duplicates while preserving order
def get_unique(iterable):
    s = set()
    return [e for e in iterable if e not in s and s.add(e) is None]

print('Unique:', get_unique([1, 2, 3, 1, 2, 3, 4, 1]))
````

## 6.2 Set and Dictionary Comprehensions

```python
{expr for var in iterable}                  # set comprehension

```

````python
# Set comprehension — automatic deduplication
sentences = ['Warsaw Paris Berlin', 'Praha Warsaw Berlin Caen']
unique_words = {word for s in sentences for word in s.split()}
print('Unique words:', unique_words)

# Common elements of 3 lists
l1, l2, l3 = [1,3,7,4], [3,4,5,7], [4,5,6,7]
common = {x for x in l1 if x in l2 and x in l3}
print('Common:', common)

# Dict comprehension
nums = [1, 2, 3, 4]
parity = {n: ('odd' if n%2 else 'even') for n in nums}
print('Parity:', parity)

# Letter counts
text = 'to be or not to be'
letter_counts = {c: text.count(c) for c in set(text) if c.isalpha()}
print('Letter counts:', dict(sorted(letter_counts.items())))

# Score → grade
scores = {'Eve': 83, 'Bob': 47, 'Sue': 65, 'Dave': 77}
grades = {name:
           '5'  if score >= 90 else
           '4+' if score >= 80 else
           '4'  if score >= 70 else
           '3+' if score >= 60 else
           '3'  if score >= 50 else '2'
           for name, score in scores.items()}
print('Grades:', grades)

# Reverse a dictionary
dct = {'Mary': 'John', 'Kate': 'Bill', 'Sophia': 'Kevin'}
print('Reversed:', {v: k for k, v in dct.items()})

# Filter a dictionary
d = {'a': 9, 'b': 12, 'c': 7, 'd': 14}
print('Filtered (<=10):', {k: v for k, v in d.items() if v <= 10})
````

## 6.3 Generator Functions

A function with `yield` is a **generator function**. Calling it returns a **generator object** — no code runs yet!

Key properties:
- **Lazy**: computes values on demand, one at a time
- **Memory-efficient**: doesn't store all values
- Can represent **infinite** sequences
- `yield` suspends execution and returns a value; resumed on next `next()` call
- `return` in a generator raises `StopIteration`
- **Cannot be restarted** — once exhausted, must create new generator

````python
import sys

# Generator vs list — memory comparison
N = 1_000_000
lst = [x**2 for x in range(N)]
gen = (x**2 for x in range(N))    # generator expression
print(f'List size:      {sys.getsizeof(lst):,} bytes')
print(f'Generator size: {sys.getsizeof(gen):,} bytes')

# Fibonacci generator
def fibonacci(n):
    a, b, count = 0, 1, 0
    while count <= n:
        yield a
        a, b = b, a + b
        count += 1

fib_gen = fibonacci(14)
print('\nFibonacci(14):', list(fib_gen))
print('Exhausted — iterate again:', list(fibonacci(14)))  # must create new!

# Infinite generator — squares
def squares():
    i = 1
    while True:
        yield i * i
        i += 1

gen2 = squares()
print('\nFirst 10 squares:', [next(gen2) for _ in range(10)])

# Generator of generators — limiter
from random import randint

def limiter(gen_func, n):
    g = gen_func()
    for _ in range(n):
        yield next(g)

def dice():
    while True:
        yield randint(1, 6)

print('\n5 Fibonacci:', list(limiter(lambda: fibonacci(100), 5)))
print('5 dice rolls:', list(limiter(dice, 5)))
````

## 6.4 Generator Expressions

Like list comprehensions but with `()` → creates a generator (lazy, memory-efficient):

```python
gen = (x**2 for x in range(10))    # generator expression
lst = [x**2 for x in range(10)]    # list comprehension — all values in memory
```

Generator expressions are often used directly in function calls: `sum(x**2 for x in range(n))`

````python
# Sum without creating a list
total = sum(i*i for i in range(1, 1_000_001))
print(f'Sum of squares 1..1M = {total:,}')

# Using generator expression in max()
lst = [3, 5, 2, 3, 5, 3, 1, 3, 6]
most_common = max(set(lst), key=lst.count)
print(f'Most common in {lst}: {most_common}')

# Generator for prime check (all() is lazy — stops on first False)
from math import isqrt
def is_prime(n):
    if n < 2: return False
    return all(n % m != 0 for m in range(2, isqrt(n)+1))

primes = [n for n in range(2, 50) if is_prime(n)]
print('Primes <50:', primes)

# any() and all() with generator expressions
nums = [2, 4, 6, 8, 10]
print('All even?', all(n%2==0 for n in nums))   # True
print('Any >7?',   any(n>7   for n in nums))    # True
````

## 6.5 Communicating with Generators (Coroutines)

Generators can **receive** values via `gen.send(value)`:
1. `next(gen)` or `gen.send(None)` — **prime** the generator (run to first `yield`)
2. `gen.send(v)` — resumes generator; `v` becomes the value of `x` in `x = yield expr`
3. `gen.close()` — closes generator (raises `GeneratorExit` inside it)

Pattern: use a **`@coroutine` decorator** that primes automatically (calls `next()` once)

````python
import functools

# Priming decorator
def coroutine(func):
    @functools.wraps(func)
    def primed(*args, **kwargs):
        gen = func(*args, **kwargs)
        next(gen)   # prime it
        return gen
    return primed

# A simple coroutine that receives values and accumulates stats
@coroutine
def statistics():
    count = total = 0
    while True:
        value = yield (count, total, total/count if count else 0)
        count += 1
        total += value

stat = statistics()
for n in [10, 20, 30, 40, 50]:
    result = stat.send(n)
    cnt, tot, avg = result
    print(f'After sending {n:3}: count={cnt}, total={tot}, avg={avg:.1f}')

stat.close()

print()
# Two-way generator (from lecture)
def twoWayGener():
    value = None
    while True:
        x = yield value     # receive x, send back value
        value = 2 * x       # double what we received

gen = twoWayGener()
next(gen)               # prime
print(gen.send(1))      # sends 1, receives 2
print(gen.send(3))      # sends 3, receives 6
print(gen.send(7))      # sends 7, receives 14
````

## 6.6 `itertools` — Important Tools

| Function | Description |
|----------|-------------|
| `accumulate(iter, func)` | Running totals (all intermediate results) |
| `dropwhile(pred, iter)` | Skip elements while pred is True, then yield rest |
| `takewhile(pred, iter)` | Yield elements while pred is True, then stop |
| `islice(iter, stop)` | First `stop` elements |
| `islice(iter, start, stop, step)` | Like range but for any iterable |
| `count(start, step)` | Infinite counter (like range without stop) |
| `groupby(iter, key)` | Group consecutive equal-key elements |
| `filterfalse(pred, iter)` | Elements where pred is False |

⚠️ `groupby` requires the input to be **sorted** by the same key first!

````python
from itertools import (
    dropwhile, takewhile, islice,
    count, groupby, filterfalse, accumulate
)

ls = [1, 2, 3, 4, 5, 6]

# dropwhile — skip while condition holds, then yield rest
print('dropwhile(<4):', list(dropwhile(lambda x: x < 4, ls)))  # [4,5,6]

# takewhile — yield while condition holds, then stop
print('takewhile(<5):', list(takewhile(lambda x: x < 5, ls)))  # [1,2,3,4]

# islice — like slice for any iterable
s = '0123456789'
print('islice first 5:', list(islice(s, 5)))          # ['0','1','2','3','4']
print('islice 2..8:   ', list(islice(s, 2, 8)))       # ['2'..'7']
print('islice every 3:', list(islice(s, None, None, 3)))# ['0','3','6','9']

# count — infinite counter
cnt = count(1.5, 2.5)
print('count(1.5,2.5):', list(islice(cnt, 6)))  # [1.5, 4.0, 6.5, 9.0, 11.5, 14.0]

# Primes using islice (first 10)
from sys import maxsize
all_primes = (k for k in range(2, maxsize)
              if all(k%m != 0 for m in range(2, int(k**0.5)+1)))
print('First 10 primes:', list(islice(all_primes, 10)))

print()
# groupby — must sort first by same key!
words = ['frog', 'goat', 'France', 'Greg', 'floor']
key = lambda s: s.lower()[0]
words.sort(key=key)
for letter, group in groupby(words, key=key):
    print(f"Group '{letter}': {list(group)}")

print()
# filterfalse — opposite of filter
nums = [1, 2, 3, 4, 5, 6]
print('Odd only (filterfalse even):', list(filterfalse(lambda x: x%2==0, nums)))
````

# 🎯 Test Preparation — Key Points Summary

## Must Know for the Test

**Objects & References**
- `id()` = memory address; `type()` = type; `is` = same object; `==` = same value
- Immutable types create new objects on 'change'; mutable types modify in-place
- Shallow copy (`list(a)`, `a[:]`) vs deep copy (`copy.deepcopy(a)`)

**Flow Control**
- `else` on loops = runs only if NO `break`
- `match` = no fall-through, `case _:` = catch-all, can match types and values with guards

**Types**
- `/` always returns `float`; `//` is floor division (towards -∞, not towards 0!)
- `bool` inherits from `int`: `True==1`, `False==0`, `True/3 == 0.333...`
- `int` has unlimited precision; small ints cached in `[-5, 256]`
- `0.1` is NOT exact; only powers-of-2 fractions are exact in `float`

**Functions**
- LEGB rule; `global`, `nonlocal` keywords
- Default values created ONCE — mutable defaults are dangerous
- `*args` → tuple; `**kwargs` → dict
- `and`/`or` return an operand, not `True`/`False`
- `:=` walrus assigns AND returns value, needs parentheses

**Decorators**
- `@decorator` ≡ `func = decorator(func)`
- Always use `@functools.wraps(func)` to preserve metadata
- Decorator factory: `@rangecheck(0,255)` ≡ `func = rangecheck(0,255)(func)`

**Comprehensions**
- `[expr for x in it if cond]` → list; `{expr...}` → set; `{k:v ...}` → dict; `(expr...)` → generator
- Loop variable is local (doesn't leak)
- Nested comprehension order mirrors nested for-loops

**Generators**
- `yield` suspends; `next()` resumes; `StopIteration` when exhausted
- Cannot be restarted; cannot index/slice
- `send(v)` → v becomes value of `x = yield expr`; must prime first with `next()`
- `itertools`: `dropwhile`, `takewhile`, `islice`, `count`, `groupby`, `accumulate`
