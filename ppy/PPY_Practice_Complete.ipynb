### ✍ WRITE: EvenNumbers(max_val) iterator

````python
class EvenNumbers:
    def __init__(self, max_val):
        self.current = 0
        self.max_val = max_val

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= self.max_val:
            raise StopIteration
        val = self.current
        self.current += 2
        return val

for n in EvenNumbers(10):
    print(n, end=' ')   # 0 2 4 6 8
````

## 12 · Two-Way Generators (send)

````python
def twoWayGener():
    print('Gene: started')
    value = None
    while True:
        print(f'Gene: ready, {value=}')
        x = yield value
        print(f'Gene: resumed, {x=}')
        value = 2 * x

gen = twoWayGener()
print('Main: priming')
ret = next(gen)          # prime — runs until first yield
print(f'Main: primed: {ret=}')

ret = gen.send(1)
print(f'Main: got {ret=}')

ret = gen.send(3)
print(f'Main: got {ret=}')
````

````python
# Priming decorator
import functools

def coroutine(func):
    @functools.wraps(func)
    def primed(*args, **kwargs):
        gen = func(*args, **kwargs)
        next(gen)    # auto-prime
        return gen
    return primed

@coroutine
def longlines(n):
    while True:
        line = yield
        if len(line.split()) > n:
            print('Long:', line)

gen = longlines(3)   # already primed
gen.send('one two')
gen.send('one two three four five')
gen.send('hi')
````

### ✍ WRITE: running_total coroutine

````python
import functools

def coroutine(func):
    @functools.wraps(func)
    def primed(*args, **kwargs):
        p = func(*args, **kwargs)
        next(p)
        return p
    return primed

@coroutine
def running_total():
    total = 0
    while True:
        val = yield total
        total += val

gen = running_total()
print(gen.send(5))    # 5
print(gen.send(3))    # 8
print(gen.send(10))   # 18
````

````python
# functools.wraps effect
import functools

def deco_without_wraps(func):
    def wrapper(*args, **kwargs): return func(*args, **kwargs)
    return wrapper

def deco_with_wraps(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs): return func(*args, **kwargs)
    return wrapper

@deco_without_wraps
def f(): 'my docstring'

@deco_with_wraps
def g(): 'my docstring'

print(f.__name__, '|', f.__doc__)         # ?
print(g.__name__, '|', g.__doc__)         # ?
print(hasattr(g, '__wrapped__'))          # ?
print(g.__wrapped__ is g)                 # ?
````

### ✍ WRITE: decorators

````python
import functools, time

# 1. timer decorator
def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        t0     = time.perf_counter()
        result = func(*args, **kwargs)
        print(f'{func.__name__}: {time.perf_counter()-t0:.4f}s')
        return result
    return wrapper

@timer
def slow():
    time.sleep(0.05)
    return 42

slow()

# 2. call counter
def counting(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        wrapper.counter += 1
        return func(*args, **kwargs)
    wrapper.counter = 0
    return wrapper

@counting
def add(a, b): return a + b

add(1, 2); add(3, 4); add(5, 6)
print(add.counter)   # 3
````

````python
# 3. rangecheck decorator factory
import functools

def rangecheck(mn, mx):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for v in list(args) + list(kwargs.values()):
                if isinstance(v, (int, float)) and not (mn <= v <= mx):
                    raise ValueError(f'{v} out of range [{mn}, {mx}]')
            return func(*args, **kwargs)
        return wrapper
    return decorator

@rangecheck(0, 255)
def color(r, g, b):
    return f'rgb({r},{g},{b})'

print(color(100, 200, 50))
try:
    print(color(0, 300, 0))
except ValueError as e:
    print(e)
````

````python
# 4. Singleton class decorator
import functools

def Singleton(cls):
    instances = {}
    @functools.wraps(cls)
    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]
    return get_instance

@Singleton
class Config:
    def __init__(self): self.debug = False

c1, c2 = Config(), Config()
print(c1 is c2)   # True
````

### ✍ WRITE: BankAccount with property

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self._owner   = owner
        self._balance = float(balance)

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0: raise ValueError('Deposit must be positive')
        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0: raise ValueError('Withdrawal must be positive')
        if amount > self._balance: raise ValueError('Insufficient funds')
        self._balance -= amount

    def __repr__(self):
        return f"BankAccount('{self._owner}', balance={self._balance:.2f})"

acc = BankAccount('Alice', 100)
acc.deposit(50)
acc.withdraw(30)
print(acc)           # BankAccount('Alice', balance=120.00)
print(acc.balance)   # 120.0
````

### ✍ WRITE: Vector2D

````python
from math import sqrt

class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):  return Vector2D(self.x + other.x, self.y + other.y)
    def __sub__(self, other):  return Vector2D(self.x - other.x, self.y - other.y)
    def __mul__(self, scalar): return Vector2D(self.x * scalar,  self.y * scalar)
    def __rmul__(self, scalar): return self.__mul__(scalar)
    def __eq__(self, other):   return (self.x, self.y) == (other.x, other.y)
    def __hash__(self):        return hash((self.x, self.y))
    def __abs__(self):         return sqrt(self.x**2 + self.y**2)
    def __repr__(self):        return f'Vector2D({self.x}, {self.y})'

v1 = Vector2D(1, 2)
v2 = Vector2D(3, 4)
print(v1 + v2)           # Vector2D(4, 6)
print(v1 - v2)           # Vector2D(-2, -2)
print(v1 * 3)            # Vector2D(3, 6)
print(3 * v1)            # Vector2D(3, 6)
print(abs(Vector2D(3,4)))   # 5.0
print(v1 == Vector2D(1,2))  # True
````

## 17 · Exceptions

````python
def fun(num, denom):
    try:
        print('In try')
        z = num // denom
    except ArithmeticError as e:
        print('Exception:', e)
        return None
    else:
        print('Result:', z)
        return z
    finally:
        print('finally — always runs')

print(fun(27, 4))
print('---')
print(fun(31, 0))
````

````python
# Exception as multi-loop break
lst    = [9, 1, 5, 7, 11, 10, 4]
target = 20

try:
    for a in range(len(lst)-2):
        for b in range(a+1, len(lst)-1):
            for c in range(b+1, len(lst)):
                if lst[a]+lst[b]+lst[c] == target:
                    raise Exception(a, b, c)
except Exception as ex:
    result = ex.args
else:
    result = (-1, -1, -1)

print(result)
print([lst[i] for i in result])   # verify the triplet
````

### ✍ WRITE: custom exception

````python
class OutOfRange(ValueError):
    def __init__(self, val, lower, upper):
        super().__init__(f'{val} is not in range [{lower}, {upper}]')
        self.val   = val
        self.lower = lower
        self.upper = upper

def get_month(n):
    names = ['Jan','Feb','Mar','Apr','May','Jun',
             'Jul','Aug','Sep','Oct','Nov','Dec']
    if not (1 <= n <= 12):
        raise OutOfRange(n, 1, 12)
    return names[n - 1]

try:
    print(get_month(13))
except OutOfRange as e:
    print(e)

print(get_month(5))   # May
````

## 18 · Context Managers

````python
import time

class TimeIt:
    def __enter__(self):
        self.start = time.perf_counter()
        return self    # 'as' variable gets this

    def __exit__(self, exc_type, exc_val, trace):
        self.time = time.perf_counter() - self.start
        return False   # don't suppress exceptions

with TimeIt() as tim:
    total = sum(range(1_000_000))

print(f'{tim.time:.4f} sec')
print(total)
````

````python
import contextlib

@contextlib.contextmanager
def atBeginning(lst):
    lst.reverse()
    yield
    lst.reverse()

lst = [3, 4, 5]
print('Before:', lst)
with atBeginning(lst):
    lst.append(2)
    lst.append(1)
    print('Inside:', lst)    # ?
print('After:', lst)         # ?
````

### ✍ WRITE: SuppressErrors context manager

````python
class SuppressErrors:
    def __init__(self, *exceptions):
        self.exceptions = exceptions

    def __enter__(self):
        return self

    def __exit__(self, exc_type, exc_value, tb):
        if exc_type is not None and issubclass(exc_type, self.exceptions):
            return True   # suppress
        return False      # propagate

with SuppressErrors(ZeroDivisionError):
    x = 1 / 0
print('Continues after suppressed exception')

try:
    with SuppressErrors(ValueError):
        raise TypeError('should propagate')
except TypeError:
    print('TypeError correctly propagated')
````

````python
# @contextmanager version — managed decimal precision
import contextlib, decimal

@contextlib.contextmanager
def managed_precision(n):
    ctx = decimal.getcontext()
    old = ctx.prec
    ctx.prec = n
    try:
        yield
    finally:
        ctx.prec = old

print('Before:', decimal.Decimal(7) / decimal.Decimal(3))
with managed_precision(4):
    print('Inside:', decimal.Decimal(7) / decimal.Decimal(3))
print('After: ', decimal.Decimal(7) / decimal.Decimal(3))
````

## 20 · functools: singledispatch, total_ordering

````python
from functools import singledispatch

@singledispatch
def process(arg):
    return f'unknown: {type(arg).__name__}'

@process.register(int)
def _(v): return f'int: {v * 2}'

@process.register(str)
def _(s): return f'str: {s.upper()}'

@process.register(list)
def _(seq): return f'seq: {list(seq)}'

print(process(5))        # ?
print(process('hello'))  # ?
print(process([1,2,3]))  # ?
print(process(True))     # ?  bool dispatches to int!
print(process(3.14))     # ?  no handler for float
````

````python
from functools import total_ordering

@total_ordering
class Student:
    def __init__(self, name, grade):
        self.name  = name
        self.grade = grade

    def __eq__(self, other):
        return self.grade == other.grade

    def __lt__(self, other):
        return self.grade < other.grade

    def __repr__(self):
        return f'Student({self.name},{self.grade})'

students = [Student('Alice',85), Student('Bob',72), Student('Charlie',91)]
print(sorted(students))               # sorted by grade
print(max(students))                  # ?
print(students[0] >= students[1])     # ?  generated by total_ordering
````

````python
# 3. Decorator timing and __wrapped__
import functools

def deco(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@deco
def adder(a, b): return a + b

print(adder.__name__)             # ?
print(adder.__wrapped__(3, 4))    # ?  call original via __wrapped__
print(adder(3, 4))                # ?
````

````python
# 4. MRO + cooperative super()
class A:
    def hello(self):
        print('A.hello')

class B(A):
    def hello(self):
        print('B.hello')
        super().hello()

class C(A):
    def hello(self):
        print('C.hello')
        super().hello()

class D(B, C):
    def hello(self):
        print('D.hello')
        super().hello()

D().hello()
print(D.__mro__)
````
