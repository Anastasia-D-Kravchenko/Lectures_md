## 🧠 Theory

### Python REPL
Python's interactive shell (REPL = Read-Eval-Print-Loop) lets you type single commands and see results immediately.
```
python        # start in terminal
>>> x = 10
>>> type(x)   # <class 'int'>
>>> exit()    # quit
```

### Variables and Types
Python is dynamically typed – every value carries a type, but you don't declare it.
```python
x = 42        # int
y = 3.14      # float
s = "hello"   # str
flag = True   # bool
```

### Type conversion
```python
int("42")     # -> 42
float("3.14") # -> 3.14
str(99)       # -> "99"
bool(0)       # -> False   (0, "", [], None are falsy)
```

### Arithmetic operators
| Operator | Meaning          | Example |
|----------|------------------|---------|
| `+`      | Addition         | `5+3 -> 8` |
| `-`      | Subtraction      | `5-3 -> 2` |
| `*`      | Multiplication   | `5*3 -> 15` |
| `/`      | Division (float) | `5/2 -> 2.5` |
| `//`     | Integer division | `5//2 -> 2` |
| `%`      | Modulo           | `5%2 -> 1` |
| `**`     | Exponentiation   | `2**8 -> 256` |

### Logical operators
`and`, `or`, `not`

### String essentials
```python
s = "Hello World"
len(s)              # 11
```

## ✏️ Exercise III - String operations

````python
# Exercise III - explore operations on a string variable
text = "Hello"

# string + string
print(text + " World")

# string * int
print(text * 3)

# string + int -> would raise TypeError
try:
    print(text + 42)
except TypeError as e:
    print(f"Error: {e}")

# cast to int and do arithmetic
num = int("10")
print(num + 5)
print(num * 3)
print(num / 2)
print(type(num / 2))    # float
````

## ✏️ Exercise IV - Division vs integer division

````python
# Exercise IV - float division vs integer division
a = 10
b = 3

print(a / b)          # 3.333... -> float
print(a // b)         # 3        -> int
print(type(a / b))    # <class 'float'>
print(type(a // b))   # <class 'int'>
# Note: / always returns float, // truncates toward negative infinity
````

## ✏️ Exercise V - Simple calculator

````python
# Exercise V - calculator
a = 15
b = 4

print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Division:", a / b)
print("Integer division:", a // b)
print("Exponentiation:", a ** b)

print(f"Sum = {a+b}, Diff = {a-b}, Product = {a*b}")
print(f"/ = {a/b:.4f}, // = {a//b}, ** = {a**b}")
````

## ✏️ Exercise VI - Swapping variables

````python
# Exercise VI - two ways to swap
a = 5
b = 9

# Classic (temp variable)
tmp = a
a = b
b = tmp
print("Classic swap:", a, b)   # 9 5

a, b = 5, 9

# Pythonic (tuple unpacking)
a, b = b, a
print("Tuple swap:", a, b)     # 9 5
````

## ✏️ Exercise VII - Conditional checks

````python
# Exercise VII - positive/even/negative checks
n = -14

print("Positive and even:", n > 0 and n % 2 == 0)
print("Negative or > 100:", n < 0 or n > 100)
print("Not positive:", not (n > 0))
````

## ✏️ Exercise VIII - Integer representations

````python
# Exercise VIII - different bases
n = 255

print("Binary:", bin(n))   # 0b11111111
print("Octal:", oct(n))    # 0o377
print("Hex:", hex(n))      # 0xff

# Parsing strings in different bases
print(int("1111", 2))   # binary -> 15
print(int("FF", 16))    # hex    -> 255
print(int("17", 8))     # octal  -> 15
````

## ✏️ Exercise IX - Big integers

````python
# Python int has unlimited precision - no overflow!
result = 2 ** 200
print(result)
print(type(result))
````

## ✏️ Exercise X - Immutability of integers

````python
# Numbers are immutable; variables are just labels
x = 10
print("id before:", id(x))
x = x + 1
print("id after:", id(x))
# id changes because int objects are immutable;
# x+1 creates a NEW int object, x is rebound to it.
````

## ✏️ Exercise XI - String operations

````python
# Exercise XI - string methods
sentence = "The quick brown fox jumps over the lazy dog"

print("Length:", len(sentence))
print("With underscores:", sentence.replace(" ", "_"))
print("Words:", sentence.split())
print("First 3 chars:", sentence[:3])
print("Last 4 chars:", sentence[-4:])
````

## ✏️ Exercises XII-XVI - String analysis

````python
text = "Python is great. I love Python programming 123 times."

# XII - word count
words = text.split()
print("Word count:", len(words))

# XIII - digit count
digits = sum(c.isdigit() for c in text)
print("Digits:", digits)

# XIV - isdigit check
print("'123'.isdigit():", "123".isdigit())

# XV - isalpha check
print("'Hello'.isalpha():", "Hello".isalpha())

# XVI - contains, count, replace
print("Contains 'Python':", "Python" in text)
print("Count of 'a':", text.count("a"))
print("After replace:", text.replace("Python", "JAVA"))
````

## ✏️ Exercise XVII - Boolean values

````python
# Exercise XVII - truthy / falsy
for val in [1, 0, 100, "", "Ala ma kota", 0.5]:
    print(f"bool({val!r}) -> {bool(val)}")
````

## ✏️ Exercise XVIII - Number analysis

````python
# Exercise XVIII - combined number analysis
n = -42

sign = "positive" if n > 0 else ("negative" if n < 0 else "zero")
print("Sign:", sign)
print("Even:", n % 2 == 0)
print("Binary:", bin(n))
print("Hex:", hex(n))
print("Square:", n ** 2)
````
