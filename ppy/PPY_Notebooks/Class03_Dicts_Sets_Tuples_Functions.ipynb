## 🧠 Theory

### Dictionary
```python
d = {"key": "val", "age": 25}
d["key"]              # access
d.get("x", default)  # safe access
d["new"] = 99         # add / update
del d["key"]          # delete
d.keys() / d.values() / d.items()
"key" in d            # membership
{k: v for k, v in pairs}  # dict comprehension
d.setdefault(key, default) # insert if missing
```

### Set
```python
s = {1, 2, 3}
s.add(x) / s.discard(x)   # discard: no error if missing
s | t   # union         s & t   # intersection
s - t   # difference    s ^ t   # symmetric difference
```

### Tuple
```python
t = (1, 2, 3)
t[0]           # access (read-only – immutable)
a, b, c = t    # unpacking
(a, *rest) = t # starred unpacking
```

### Functions
```python

lambda x: x**2                       # anonymous

def total(*args): return sum(args)   # variadic positional
def info(**kwargs):                  # variadic keyword

```

## ✏️ Task 1 - Shopping cart (dictionary)

````python
cart = {}

def add_product(name, price):
    cart[name] = price

items = [("Apple",0.99),("Milk",1.49),("Bread",2.59),("Butter",3.29),("Eggs",2.09)]
for name, price in items:
    add_product(name, price)

total    = sum(cart.values())
priciest = max(cart, key=cart.get)
avg      = total / len(cart)

print("=== Shopping Cart ===")
for name, price in cart.items():
    print(f"  {name:<10} ${price:.2f}")
print(f"Total:          ${total:.2f}")
print(f"Most expensive: {priciest} (${cart[priciest]:.2f})")
print(f"Average price:  ${avg:.2f}")
````

## ✏️ Task 2 - User management (set)

````python
users = set()

def add_user(name):
    if name in users:
        print(f"  '{name}' already exists")
    else:
        users.add(name)

for name in ["Alice","Bob","Charlie","Alice","Dave","Eve","Bob"]:
    add_user(name)

sorted_users = sorted(users)
first_three  = tuple(sorted_users[:3])

print("All users (sorted):", sorted_users)
print("Unique count:      ", len(users))
print("First 3 (tuple):   ", first_three)
print("Longest name:      ", max(users, key=len))
for name in ["Alice", "Zara"]:
    print(f"  '{name}' exists: {name in users}")
````

## ✏️ Task 3 - Rock-Paper-Scissors (dict + tuple history)

````python
import random

wins_over = {"paper": "rock", "rock": "scissors", "scissors": "paper"}
moves = list(wins_over.keys())

def get_result(user, comp):
    if user == comp: return "draw"
    return "win" if wins_over[user] == comp else "loss"

history  = []
score    = {"win": 0, "loss": 0, "draw": 0}
user_seq = ["rock", "paper", "scissors"]

for rnd, user_move in enumerate(user_seq, 1):
    comp_move = random.choice(moves)
    result    = get_result(user_move, comp_move)
    score[result] += 1
    history.append((user_move, comp_move, result))
    print(f"Round {rnd}: you={user_move:8} comp={comp_move:8} -> {result}")

print("\nRound history:", history)
print("Score:        ", score)
````

## ✏️ Task 4 - Filter with predicate function

````python
N   = 15
lst = list(range(5, 5 + N * 4, 4))

def is_div3(n): return n % 3 == 0
def is_odd(n):  return n % 2 != 0

def apply_filter(lst, predicate):
    return [x for x in lst if predicate(x)]

print("Original:", lst)
print("Div by 3:", apply_filter(lst, is_div3))
print("Odd:     ", apply_filter(lst, is_odd))
print("> 30:    ", apply_filter(lst, lambda x: x > 30))
````

## ✏️ Task 5 - Name lengths dictionary

````python
names = ["John","Mary","Kitty","Alice","John","Mary","Al","Bill","John","Alice","Al"]

by_len_list = {}
by_len_set  = {}
count_dict  = {}

for name in names:
    l = len(name)
    by_len_list.setdefault(l, []).append(name)
    by_len_set.setdefault(l, set()).add(name)
    count_dict[name] = count_dict.get(name, 0) + 1

print("Lists by length:", by_len_list)
print("Sets by length: ", by_len_set)
print("Occurrences:    ", count_dict)
````

## ✏️ Task 6 - Collect N distinct numbers

````python
stream = [3, 7, 3, 2, 7, 9, 1, 2, 5, 3, 8]
N = 5

seen     = set()
repeated = set()

for num in stream:
    if num in seen:
        repeated.add(num)
    seen.add(num)
    if len(seen) == N:
        break

print(f"Collected {N} distinct: {sorted(seen)}")
print(f"Repeated numbers:       {sorted(repeated)}")
````

## ✏️ Task 7 - Chess tournament rankings

````python
dct = {"John":3,"Bill":4,"Jane":4,"Kim":2,"Mary":3,"Joe":0,"Sue":5,"Ada":2,"Ray":2}

by_wins = {}
for player, wins in dct.items():
    by_wins.setdefault(wins, []).append(player)

print("Wins -> players:", by_wins)

ranking = sorted(by_wins.items(), key=lambda x: x[0], reverse=True)
print("\nRanking (descending):")
for wins, players in ranking:
    print(f"  {wins} wins: {sorted(players)}")
````

## ✏️ Functions – *args and **kwargs

````python
def sum_all(*args):
    return sum(args)

def multiply_all(*args):
    result = 1
    for x in args: result *= x
    return result

def print_user(**kwargs):
    for k, v in kwargs.items():
        print(f"  {k}: {v}")

def stats_calculator(*args, round_digits=2, **kwargs):
    total = sum(args)
    avg   = total / len(args) if args else 0
    return {"sum": round(total, round_digits), "average": round(avg, round_digits)}

print("sum_all(1,2,3,4):", sum_all(1,2,3,4))
print("multiply_all(2,3,4):", multiply_all(2,3,4))
print("User info:")
print_user(name="Anna", age=22, city="Warsaw")
print("Stats:", stats_calculator(4,8,15,16,23,42, round_digits=1))
````

## ✏️ Set & Tuple utility functions

````python
def unique_names(names):        return set(names)
def common_elements(s1, s2):    return s1 & s2
def all_unique(numbers):        return len(numbers) == len(set(numbers))
def set_operations(s1, s2):
    return {"union": s1|s2, "intersection": s1&s2,
            "difference": s1-s2, "sym_diff": s1^s2}

def remove_duplicates(lst):
    seen = set()
    return [x for x in lst if not (x in seen or seen.add(x))]

names = ["Alice","Bob","Alice","Charlie","Bob","Dave"]
print("Unique:", sorted(unique_names(names)))
print("Common:", common_elements({1,2,3,4},{3,4,5,6}))
print("All unique [1,2,3]:", all_unique([1,2,3]))
print("All unique [1,2,2]:", all_unique([1,2,2]))
print("Ops:", set_operations({1,2,3},{2,3,4}))
print("No dups:", remove_duplicates([1,3,2,1,4,3,5]))
````
