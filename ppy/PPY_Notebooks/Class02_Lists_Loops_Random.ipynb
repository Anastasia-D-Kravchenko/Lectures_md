## 🧠 Theory

### Lists
```python
lst = [1, 2, 3, 4, 5]
lst.append(6)        # add at end
lst.insert(0, 0)     # insert at index
lst.remove(3)        # remove first occurrence of 3
lst.pop()            # remove & return last element
lst.sort()           # sort in-place
sorted(lst)          # return new sorted list
lst[::-1]            # reversed copy
lst[2:5]             # slice [2,3,4]
lst[::2]             # every 2nd element
lst[:]               # shallow copy
```

### Random module
```python
import random
random.randint(1, 100)    # random int in [1, 100]
random.choice(lst)        # random element from list
random.shuffle(lst)       # shuffle in-place
```

### List comprehension
```python
squares = [x**2 for x in range(10)]
evens   = [x for x in lst if x % 2 == 0]
```

### Loops
```python
for item in lst: ...
for i in range(5): ...                # 0,1,2,3,4
for i, v in enumerate(lst): ...      # index + value
while condition:
    break      # exit loop immediately
    continue   # skip to next iteration
# else on loops: runs if loop finished without break
```

## ✏️ Task 1 - Student score grader

````python
import random

scores = [random.randint(0, 100) for _ in range(50)]

def assign_grade(score):
    if score >= 90: return 5
    elif score >= 75: return 4
    elif score >= 60: return 3
    else: return 2

grades  = [assign_grade(s) for s in scores]
average = sum(scores) / len(scores)
passed  = sum(1 for s in scores if s >= 60)
top5    = sorted(scores, reverse=True)[:5]

print(f"Average score:   {average:.2f}")
print(f"Students passed: {passed}/{len(scores)}")
print(f"Top 5 scores:    {top5}")
print(f"Grade sample:    {grades[:10]}")
````

## ✏️ Task 2 - Number guessing game (binary-search simulation)

````python
import random

secret   = random.randint(1, 100)
lo, hi   = 1, 100
attempts = 0
guesses  = []

print(f"(Secret for demo: {secret})")

while lo <= hi:
    guess = (lo + hi) // 2
    guesses.append(guess)
    attempts += 1
    if guess < secret:
        print(f"Attempt {attempts}: {guess} -> too small")
        lo = guess + 1
    elif guess > secret:
        print(f"Attempt {attempts}: {guess} -> too large")
        hi = guess - 1
    else:
        print(f"Attempt {attempts}: {guess} -> correct!")
        break

print(f"Total attempts: {attempts}")
print(f"Last 3 guesses: {guesses[-3:]}")
````

## ✏️ Task 3 - Numerical data analysis

````python
import random

data = [random.randint(1, 100) for _ in range(20)]
print("Data:", data)

avg            = sum(data) / len(data)
above_avg      = sum(1 for x in data if x > avg)
second_largest = sorted(data, reverse=True)[1]

print(f"Max: {max(data)}  Min: {min(data)}  Avg: {avg:.2f}")
print(f"Above average:  {above_avg}")
print(f"Even numbers:   {[x for x in data if x % 2 == 0]}")
print(f"Every 2nd el:   {data[::2]}")
print(f"Second largest: {second_largest}")
````

## ✏️ Task 5 - User management

````python
users = []

def add_user(name):
    if name in users:
        print(f"  '{name}' already exists - skipped")
    else:
        users.append(name)
        print(f"  '{name}' added")

for name in ["Alice", "Bob", "Charlie", "Alice", "Dave", "Eve", "Bob"]:
    add_user(name)

users.sort()
print("\nSorted users:", users)
print("First 5:       ", users[:5])
print("Longest name:  ", max(users, key=len))
print("'Bob' in list: ", "Bob" in users)
print("'Zara' in list:", "Zara" in users)
````

## ✏️ Task 6 - Rock-Paper-Scissors

````python
import random

moves     = ["paper", "rock", "scissors"]
wins_over = {"paper": "rock", "rock": "scissors", "scissors": "paper"}

user_seq = ["rock", "paper", "scissors"]
user_moves, comp_moves = [], []
score = {"win": 0, "loss": 0, "draw": 0}

for rnd in range(3):
    user = user_seq[rnd]
    comp = random.choice(moves)
    user_moves.append(user)
    comp_moves.append(comp)

    if user == comp:                result = "draw"
    elif wins_over[user] == comp:   result = "win"
    else:                           result = "loss"

    score[result] += 1
    print(f"Round {rnd+1}: you={user:8} comp={comp:8} -> {result}")
    if score["win"] == 2 or score["loss"] == 2:
        print("Early finish - decisive!"); break

print("\nYour moves:", user_moves)
print("Comp moves:", comp_moves)
print("Score:     ", score)
````

## ✏️ For-loop exercises (selected)

````python
# Basic loops
print("1-10:", list(range(1, 11)))
print("10-1:", list(range(10, 0, -1)))
print("Even 1-20:", [i for i in range(1, 21) if i % 2 == 0])
print("Sum 1-100:", sum(range(1, 101)))

# FizzBuzz variant
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0: print(i, "Div by 3 and 5")
    elif i % 3 == 0:               print(i, "Div by 3")
    elif i % 5 == 0:               print(i, "Div by 5")

# Prime check
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

print("Primes < 30:", [n for n in range(2, 30) if is_prime(n)])
````
