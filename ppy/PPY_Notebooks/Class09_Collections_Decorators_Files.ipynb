## ✏️ 1.1 Collections

````python
# I – count types in list
def count_types(lst):
    d = {}
    for x in lst:
        t = type(x).__name__
        d[t] = d.get(t, 0) + 1
    return d

mixed = [1, "a", 2.5, True, "b", 3, None, 2.0]
print("Types:", count_types(mixed))

# II – set of numbers > 5 from dict of lists
data = {"a": [1,6,3,8], "b": [2,9,4], "c": [5,7,11]}
big  = {n for lst in data.values() for n in lst if n > 5}
print("Numbers > 5:", sorted(big))

# III – word occurrence count
def word_count(words):
    return {w: words.count(w) for w in set(words)}

print("Word count:", word_count(["hi","bye","hi","hi","bye","ok"]))
````

## ✏️ 1.1.1 Sets

````python
# I – intersection as sorted list
a, b = [1,3,5,7,9], [3,5,6,7,8]
print("Common:", sorted(set(a) & set(b)))

# II – unique chars, no spaces, lowercase
s = "Hello World"
print("Unique chars:", sorted(set(s.lower().replace(" ", ""))))

# III – has duplicates
def has_duplicates(lst):
    return len(lst) != len(set(lst))

print("Dups [1,2,3]:", has_duplicates([1,2,3]))
print("Dups [1,2,2]:", has_duplicates([1,2,2]))

# IV – symmetric difference count
l1 = ["apple","banana","cherry"]
l2 = ["banana","date","cherry","fig"]
sym = set(l1) ^ set(l2)
print("Sym diff:", sym, "count:", len(sym))
````

## ✏️ 1.1.2 Dictionaries

````python
# I – group by first letter
words = ["apple","avocado","banana","cherry","apricot","blueberry"]
by_letter = {}
for w in words:
    by_letter.setdefault(w[0], []).append(w)
print("By letter:", by_letter)

# II – swap keys/values
d = {"a": 1, "b": 2, "c": 3}
print("Swapped:", {v: k for k, v in d.items()})

# III – sum of lists
d2 = {"x": [1,2,3], "y": [4,5], "z": [6,7,8,9]}
print("Sums:", {k: sum(v) for k, v in d2.items()})

# IV – keys sorted by value descending
d3 = {"alice":90,"bob":75,"charlie":95,"dave":80}
print("Sorted by score:", sorted(d3, key=d3.get, reverse=True))
````

## ✏️ 1.1.3 Lists

````python
# I – consecutive sums
def consec_sums(lst):
    return [lst[i] + lst[i+1] for i in range(len(lst)-1)]

print("Consec sums [1,2,3,4]:", consec_sums([1,2,3,4]))

# II – palindromes
words = ["racecar","hello","level","world","noon","python"]
print("Palindromes:", [w for w in words if w == w[::-1]])

# III – rotate right
def rotate_right(lst):
    return [lst[-1]] + lst[:-1]

print("Rotate [1,2,3,4]:", rotate_right([1,2,3,4]))

# IV – remove dups preserving order
def remove_dups(lst):
    seen = set()
    return [x for x in lst if not (x in seen or seen.add(x))]

print("No dups:", remove_dups([1,3,2,1,4,3,5,2]))
````

## ✏️ 1.1.4 Tuples

````python
# I – oldest person
people = [("Alice",30),("Bob",25),("Charlie",35),("Dave",28)]
print("Oldest:", max(people, key=lambda t: t[1]))

# II – strings sorted by number
pairs = [(3,"cherry"),(1,"apple"),(2,"banana")]
print("Sorted:", [s for _, s in sorted(pairs)])

# III – only ints from mixed tuple
mixed = (1, "hello", 3.14, 42, True, None, 7)
ints  = tuple(x for x in mixed if type(x) is int)
print("Only ints:", ints)

# IV – mutable inside immutable
t = (1, [2, 3])
t[1].append(4)
print("t after append:", t)
# Works! Tuple holds a reference to the list; the list itself is mutable.
# t[0] = 99 would fail - you cannot rebind tuple elements.
````

## ✏️ 1.2 Comprehensions

````python
# List comprehension
print("Odd squares 1-10:", [x**2 for x in range(1,11) if x % 2 != 0])

nums = [-3,-1,0,2,-5,4]
print("Abs values:", [abs(x) for x in nums])

strings = ["apple","avocado","ant","banana","almond"]
print("Lengths of a-words:", [len(s) for s in strings if s.startswith("a")])

# Set comprehension
print("Remainders mod 3:", {x % 3 for x in range(1, 21)})
print("First letters:", {s[0] for s in strings})

# Dict comprehension
print("Cubes 1-5:", {n: n**3 for n in range(1, 6)})
words2 = ["hello","world","python","data"]
print("Word lengths:", {w: len(w) for w in words2})
````

## ✏️ 1.3 Functions – *args, **kwargs

````python
# I – *args sum
def sum_all(*args): return sum(args)
print("sum_all(1,2,3,4):", sum_all(1,2,3,4))

# II – **kwargs values list
def kwargs_values(**kwargs): return list(kwargs.values())
print("values:", kwargs_values(a=1, b=2, c=3))

# III – list vs copy
original = [1, 2, 3]
ref      = original        # same object
copy     = original[:]     # new list

ref[0]  = 99
copy[0] = 77

print("original:", original)   # [99,2,3] – ref changed it
print("copy:    ", copy)       # [77,2,3] – independent
````

## ✏️ 1.3.1-1.3.2 Generators & expressions

````python
# Generator functions
def even_below(limit):
    n = 0
    while n < limit:
        if n % 2 == 0: yield n
        n += 1

print("Evens < 20:", list(even_below(20)))

def char_gen(s):
    for c in s: yield c

print("Chars:", list(char_gen("Python")))

# Generator expressions
print("Sum of squares 1-10:", sum(x**2 for x in range(1, 11)))

nums = [1, -2, 3, -4, 5, 6, 7]
print("Count > 5:", sum(1 for x in nums if x > 5))

strings = ["apple","avocado","banana","almond"]
print("Total length of 'a' words:", sum(len(s) for s in strings if s.startswith("a")))
print("Any negative:", any(x < 0 for x in nums))
print("All even [2,4,6]:", all(x % 2 == 0 for x in [2,4,6]))
print("Max string len:", max(len(s) for s in strings))

div3 = (x for x in range(1, 31) if x % 3 == 0)
print("Divisible by 3:", end=" ")
for x in div3: print(x, end=" ")
print()

char_list = ["P","y","t","h","o","n"]
print("Joined:", "".join(c for c in char_list))

nested = [[1,2,3],[4,5],[6,7,8,9]]
print("Nested sum:", sum(x for sub in nested for x in sub))
````

## ✏️ 1.3.3 Decorator exercises

````python
import functools

# I – print arguments before call
def log_args(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print(f"  Calling {func.__name__} with args={args}, kwargs={kwargs}")
        return func(*args, **kwargs)
    return wrapper

@log_args
def add(a, b): return a + b
print("add(3,4):", add(3, 4))

# II – require at least 2 positional args
def require_two_args(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        if len(args) < 2:
            print(f"  [blocked] {func.__name__} needs at least 2 args")
            return None
        return func(*args, **kwargs)
    return wrapper

@require_two_args
def multiply(a, b): return a * b
print("multiply(3,4):", multiply(3, 4))
print("multiply(3):  ", multiply(3))

# III – prefix decorator
def prefix_result(prefix):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            return f"{prefix}{func(*args, **kwargs)}"
        return wrapper
    return decorator

@prefix_result("Result: ")
def compute(x): return x * 10
print(compute(7))

# IV – call limit
def call_limit(limit):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            if wrapper.calls >= limit:
                print(f"  [blocked] {func.__name__} exceeded {limit} calls")
                return None
            wrapper.calls += 1
            return func(*args, **kwargs)
        wrapper.calls = 0
        return wrapper
    return decorator

@call_limit(3)
def say_hello(): print("  Hello!")
for _ in range(5): say_hello()

# V – type check on return value
def return_type(expected_type):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            result = func(*args, **kwargs)
            if not isinstance(result, expected_type):
                raise TypeError(f"Expected {expected_type.__name__}, got {type(result).__name__}")
            return result
        return wrapper
    return decorator

@return_type(int)
def good_func(): return 42

print("good_func():", good_func())
try:
    @return_type(int)
    def bad_func(): return "oops"
    bad_func()
except TypeError as e:
    print(f"TypeError: {e}")
````

## ✏️ 1.3.4 singledispatch recap

````python
from functools import singledispatch

@singledispatch
def smart(value):
    raise NotImplementedError(f"No handler for {type(value)}")

@smart.register(int)
def _(v): return v ** 2

@smart.register(str)
def _(v): return len(v)

@smart.register(list)
def _(v): return v[0] if v else None

print("smart(5):            ", smart(5))
print("smart('Python'):     ", smart("Python"))
print("smart([10, 20, 30]):", smart([10, 20, 30]))
````

## ✏️ 1.4 File reading

````python
import os

HOBBIT_PATH = "THE HOBBIT.txt"
for candidate in ["THE HOBBIT.txt",
                   "course_materials/Class08/THE HOBBIT.txt",
                   "/home/claude/course_materials/Class08/THE HOBBIT.txt"]:
    if os.path.exists(candidate):
        HOBBIT_PATH = candidate
        break

if os.path.exists(HOBBIT_PATH):
    def count_lines(path):
        with open(path, encoding="utf-8") as f:
            return sum(1 for _ in f)

    def lines_with_word(path, word):
        with open(path, encoding="utf-8") as f:
            return [line.strip() for line in f if word.lower() in line.lower()]

    print(f"Lines in Hobbit: {count_lines(HOBBIT_PATH)}")
    ring_lines = lines_with_word(HOBBIT_PATH, "ring")
    print(f"Lines with 'ring': {len(ring_lines)}")
    if ring_lines:
        print("Sample:", ring_lines[0][:80])
else:
    print("Hobbit file not found. Creating sample.txt instead.")
    sample = ("In a hole in the ground there lived a hobbit.\n"
              "The ring was precious to him.\n"
              "Bilbo found the ring in the dark cave.\n")
    with open("sample.txt", "w") as f:
        f.write(sample)
    print("Created sample.txt")
````
