```python
# Properties
from math import sqrt

class Square:
    def __init__(self, a):
        self._a = float(a)

    @property
    def a(self):
        return self._a

    @a.setter
    def a(self, new_a):
        if new_a < 0:
            raise ValueError('No negative sides')
        self._a = float(new_a)

    @property
    def area(self):
        return self._a ** 2

    @area.setter
    def area(self, new_area):
        if new_area < 0:
            raise ValueError('No negative areas')
        self._a = sqrt(new_area)

sq = Square(5)
print(f'a={sq.a}, area={sq.area}')   # ?
sq.area = 16
print(f'a={sq.a}, area={sq.area}')   # ?  what is sqrt(16)?
sq.a = 7
print(f'a={sq.a}, area={sq.area}')   # ?
```

## Section 14 — Collections & functools
### 🔍 READ — defaultdict

```python
from collections import defaultdict

# defaultdict auto-creates missing keys
totals = defaultdict(int)
purchases = [('Carrot', 3), ('Beetroot', 7), ('Apples', 9),
             ('Beetroot', 8), ('Apples', 8), ('Carrot', 4)]

for prod, price in purchases:
    totals[prod] += price   # no KeyError for first occurrence!

for prod, tot in totals.items():
    print(f'{prod}: {tot}')
```

```python
# singledispatch — overload by type
from functools import singledispatch

@singledispatch
def process(arg):
    return f'Unknown: {type(arg).__name__}'

@process.register(int)
def _(v):
    return f'int: {v * 2}'

@process.register(str)
def _(s):
    return f'str: {s.upper()}'

@process.register(list | tuple)
def _(arg):
    return f'seq: {list(arg)}'

print(process(5))           # ?
print(process('hello'))     # ?
print(process([1, 2, 3]))   # ?
print(process(True))        # ?  bool is subclass of int!
print(process(3.14))        # ?  no handler for float
```

**Q13: functools.wraps and print(multiply)**

```python
import functools
from functools import wraps

def deco(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@deco
def multiply(*args):
    res = 1
    for arg in args: res *= arg
    return res

print(multiply)   # ?  function name preserved by wraps
```

**Q15: iterable protocol**

```python
# Q: How is the object iterable?
# Answer: Its class has to implement the __iter__() method
# (NOT: must be a list / must have internalised tuple)

class MyIterable:
    def __iter__(self):
        return iter([1, 2, 3])

obj = MyIterable()
for x in obj:
    print(x)
```

**Q16: nums_range generator properties**

```python
def nums_range(start=0, end=0, step=1):
    while start <= end:
        yield start
        start += step

r = nums_range(0, 5)
# Select only TRUE statements:
# A. nums_range returns an iterable       → TRUE
# B. nums_range returns an iterator       → TRUE  
# C. nums_range returns a generator object → TRUE
# D. nums_range returns a function        → FALSE (it's a generator function, returns generator obj)
# E. nums_range implements iter interface  → TRUE (generator has __iter__)

print('iterable:', hasattr(r, '__iter__'))   # True
print('iterator:', hasattr(r, '__next__'))   # True
print('type:', type(r).__name__)              # generator
```
